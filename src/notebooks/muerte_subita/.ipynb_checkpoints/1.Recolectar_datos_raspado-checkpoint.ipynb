## Raspar texto desde web


#### Desde texto en pagína web mediante url específicada

```python

def extraer_texto_proyecto(url):
    """
    Extrae el texto del proyecto desde el div con la clase "container interno"
    dentro de la página web.

    Args:
        url (str): La URL de la página web.

    Returns:
        str: El texto extraído del div container interno, o None si no se encuentra.
    """
    try:
        response = requests.get(url)
        response.raise_for_status()  # Lanza una excepción para códigos de error HTTP
        soup = BeautifulSoup(response.content, 'html.parser')

        container_interno = soup.find('div', class_='container interno')
        if container_interno:
            # Extraer todo el texto dentro del contenedor, incluyendo el de los divs internos
            texto_proyecto = container_interno.get_text(separator='\n', strip=True)
            return texto_proyecto
        else:
            print(f"No se encontró el div con la clase 'container interno' en: {url}")
            return None

    except requests.exceptions.RequestException as e:
        print(f"Error al acceder a la URL {url}: {e}")
        return None
    except Exception as e:
        print(f"Ocurrió un error inesperado: {e}")
        return None

# Ejemplo de uso:
url_del_proyecto = "https://www2.hcdn.gob.ar/proyectos/textoCompleto.jsp?exp=0213-D-2020&tipo=LEY"  # Reemplaza con la URL real
texto_proyecto = extraer_texto_proyecto(url_del_proyecto)

if texto_proyecto:
    print("Texto del proyecto extraído:\n")
    print(texto_proyecto)
```

```python
# Link para descargar
pms_link_df = pms_df.loc[~pms_df['links a textos a descargar.Texto completo del proyecto'].isna(),['metadata.expediente diputados','proyecto_id','links a textos a descargar.Texto completo del proyecto']]
```

```python
# Aplicar la función a la columna 'url' del DataFrame
pms_link_df['texto_proyecto'] = pms_link_df['links a textos a descargar.Texto completo del proyecto'].apply(extraer_texto_proyecto)

# Imprimir el DataFrame resultante
pms_link_df.info()
```

```python
from PyPDF2 import PdfReader

def leer_pdf(ruta_pdf):
    """
    Lee el contenido de un archivo PDF y devuelve el texto completo.

    Args:
        ruta_pdf (str): La ruta al archivo PDF.

    Returns:
        str: El texto completo del PDF, o None si ocurre un error.
    """
    try:
        with open(ruta_pdf, 'rb') as pdf_file:
            pdf_reader = PdfReader(pdf_file)
            texto_completo = ""
            for page_num in range(len(pdf_reader.pages)):
                page = pdf_reader.pages[page_num]
                texto_completo += page.extract_text() + "\n"
            return texto_completo.strip()
    except FileNotFoundError:
        print(f"Error: El archivo PDF no se encontró en la ruta: {ruta_pdf}")
        return None
    except Exception as e:
        print(f"Ocurrió un error al leer el PDF {ruta_pdf}: {e}")
        return None

def procesar_carpeta_pdf(ruta_carpeta):
    """
    Lee todos los archivos PDF en una carpeta, extrae el texto, el ID (nombre del archivo)
    y la cantidad de páginas, y devuelve un DataFrame con esta información.

    Args:
        ruta_carpeta (str): La ruta a la carpeta que contiene los archivos PDF.

    Returns:
        pandas.DataFrame: Un DataFrame con las columnas 'id', 'texto', y 'num_paginas'.
                         Devuelve un DataFrame vacío si no se encuentran archivos PDF.
    """
    data = []
    archivos_pdf = [f for f in os.listdir(ruta_carpeta) if f.endswith('.pdf')]

    if not archivos_pdf:
        print(f"No se encontraron archivos PDF en la carpeta: {ruta_carpeta}")
        return pd.DataFrame(columns=['id', 'texto', 'num_paginas'])

    for archivo in archivos_pdf:
        ruta_completa = os.path.join(ruta_carpeta, archivo)
        try:
            with open(ruta_completa, 'rb') as pdf_file:
                pdf_reader = PdfReader(pdf_file)
                texto = ""
                for page in pdf_reader.pages:
                    texto += page.extract_text() + "\n"
                num_paginas = len(pdf_reader.pages)
                pdf_id = archivo.replace('.pdf', '')
                data.append({'id': pdf_id, 'texto': texto.strip(), 'num_paginas': num_paginas})
        except Exception as e:
            print(f"Error al procesar el archivo {archivo}: {e}")

    df = pd.DataFrame(data)
    return df

```

```python
# Especifica la ruta de la carpeta que contiene tus archivos PDF
ruta_carpeta_pdf = RESULTADOS_DIR+'/muerte_subita/pdfs/'  # Reemplaza con la ruta real

# Procesa la carpeta y obtén el DataFrame
textopdf_df = procesar_carpeta_pdf(ruta_carpeta_pdf)
```
